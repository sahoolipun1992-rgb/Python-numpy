# Session 13: NumPy Fundamentals

**NumPy (Numerical Python)** is the fundamental package for scientific computing in Python. It provides a multidimensional array object (the `ndarray`), various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays.

### 1. Importing NumPy

```python
import numpy as np
```

### 2. Creating NumPy Arrays

**a. Using `np.array()`**

```python
# 1D Array (Vector)
a1 = np.array([1, 2, 3])
print(a1)

# 2D Array (Matrix)
a2 = np.array([[1, 2, 3], [4, 5, 6]])
print(a2)

# 3D Array (Tensor)
a3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(a3)
```

**b. Using `dtype`**

```python
# Create float array
a_float = np.array([1, 2, 3], dtype=float)
print(a_float)

# Create boolean array
a_bool = np.array([0, 1, 0], dtype=bool)
print(a_bool)
```

**c. Using `np.arange()`**

```python
# np.arange(start, stop, step)
a_range = np.arange(1, 11)
print(a_range)

# With step
a_step = np.arange(1, 11, 2)
print(a_step)
```

**d. Using `np.reshape()`**

```python
# Create 10 items, reshape to 2 rows, 5 columns
a_reshape = np.arange(1, 11).reshape(2, 5)
print(a_reshape)

# 3D Reshape
a_3d_reshape = np.arange(12).reshape(2, 2, 3)
print(a_3d_reshape)
```

**e. Initialization Functions**

```python
# np.ones
print(np.ones((3, 4)))

# np.zeros
print(np.zeros((3, 4)))

# np.random
print(np.random.random((3, 4)))

# np.linspace (Linear Space)
print(np.linspace(-10, 10, 10))

# np.identity
print(np.identity(3))
```

### 3. Array Attributes

```python
a1 = np.arange(10, dtype=np.int32)
a2 = np.arange(12, dtype=float).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)

# 1. ndim
print(a3.ndim)

# 2. shape
print(a3.shape)

# 3. size
print(a3.size)

# 4. itemsize
print(a1.itemsize)
print(a2.itemsize)

# 5. dtype
print(a1.dtype)
print(a2.dtype)
```

### 4. Changing Data Types

```python
# astype
print(a3.dtype)
a3 = a3.astype(np.int32)
print(a3.dtype)
```

### 5. Array Operations

```python
a1 = np.arange(12).reshape(3, 4)
a2 = np.arange(12, 24).reshape(3, 4)

# Scalar Operations
print(a1 * 2)
print(a1 + 2)
print(a1 ** 2)

# Relational
print(a2 > 15)

# Vector Operations
print(a1 + a2)
```

### 6. Array Functions

```python
a1 = np.random.random((3, 3))
a1 = np.round(a1 * 100)

# Max/Min/Sum/Prod
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))

# Axis wise (0 = col, 1 = row)
print(np.max(a1, axis=0))
print(np.max(a1, axis=1))

# Statistics
print(np.mean(a1, axis=1))
print(np.var(a1, axis=1))
print(np.std(a1, axis=1))

# Dot Product
a = np.arange(12).reshape(3, 4)
b = np.arange(12).reshape(4, 3)
print(np.dot(a, b))
```

### 7. Indexing and Slicing

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)

# Indexing
print(a1[-1])
print(a2[1, 2])
print(a3[1, 0, 1])

# Slicing
print(a1[2:5])
print(a2[0, :])
print(a2[:, 2])
print(a2[1:, 1:3])
print(a3[1, :, 1])
```

### 8. Iterating

```python
a1 = np.arange(12).reshape(3, 4)

for i in a1:
    print(i)

for i in np.nditer(a1):
    print(i)
```

### 9. Reshaping and Transpose

```python
# Transpose
print(np.transpose(a1))
print(a1.T)

# Ravel
print(a1.ravel())
```

### 10. Stacking and Splitting

```python
a4 = np.arange(12).reshape(3, 4)
a5 = np.arange(12, 24).reshape(3, 4)

# Stacking
print(np.hstack((a4, a5)))
print(np.vstack((a4, a5)))

# Splitting
print(np.hsplit(a4, 2))
print(np.vsplit(a5, 3))
```
