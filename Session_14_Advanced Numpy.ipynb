## 1. Numpy array vs Python lists

### Speed Test

```python
# Python List Speed
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

c = []
start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])
print(f"Python List time: {time.time() - start}")
```

```python
# Numpy Array Speed
a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = time.time()
c = a + b
print(f"Numpy Array time: {time.time() - start}")
```

### Memory Test

```python
# Python List Memory
a = [i for i in range(10000000)]
print(f"Python List Memory: {sys.getsizeof(a)}")

# Numpy Array Memory
a = np.arange(10000000, dtype=np.int8) # Flexibility to change dtype
print(f"Numpy Array Memory: {sys.getsizeof(a)}")
```

## 2. Advanced Indexing

### Normal Indexing & Slicing

```python
a = np.arange(24).reshape(6,4)
print(a)

# Get specific items (Rows 1 to 2, Columns 1 to 2)
print(a[1:3, 1:3])
```

### Fancy Indexing

```python
# Getting discontinous rows (1st, 3rd, 4th, 6th)
print(a[[0, 2, 3, 5]])
```

```python
# Getting columns (1st, 3rd, 4th)
print(a[:, [0, 2, 3]])
```

### Boolean Indexing

```python
a = np.random.randint(1, 100, 24).reshape(6,4)
print(a)

# Find all numbers greater than 50
print(a[a > 50])

# Find all numbers greater than 50 and are even
print(a[(a > 50) & (a % 2 == 0)])

# Find all numbers not divisible by 7
print(a[~(a % 7 == 0)])
```

## 3. Broadcasting

The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

**Broadcasting Rules:**
1. Make the two arrays have the same number of dimensions. (Add 1 to the head of the smaller dimension array).
2. Make each dimension of the two arrays the same size. (Stretch dimensions with size 1).

```python
# Same shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print(a + b)
```

```python
# Diff shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a + b)
```

### Broadcasting Examples

```python
# Example 1
a = np.arange(12).reshape(4,3)
b = np.arange(3)
print(a + b)
```

```python
# Example 2 (Error Case)
a = np.arange(12).reshape(3,4)
b = np.arange(3)
# print(a + b) # This will throw an error because shapes (3,4) and (3,) -> (1,3) are incompatible on the last dim
```

```python
# Example 3
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)
print(a + b)
```

```python
# Example 4
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)
print(a + b)
```

## 4. Working with mathematical formulas

```python
a = np.arange(10)
print(np.sin(a))
```

### Sigmoid Function
$S(x) = \frac{1}{1 + e^{-x}}$

```python
def sigmoid(array):
    return 1 / (1 + np.exp(-array))

a = np.arange(10)
print(sigmoid(a))
```

### Mean Squared Error
$MSE = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$

```python
actual = np.random.randint(1, 50, 25)
predicted = np.random.randint(1, 50, 25)

def mse(actual, predicted):
    return np.mean((actual - predicted)**2)

print(mse(actual, predicted))
```

## 5. Working with missing values

```python
a = np.array([1, 2, 3, 4, np.nan, 6])
print(a)

# Check for missing values
print(np.isnan(a))

# Filter out missing values
print(a[~np.isnan(a)])
```
